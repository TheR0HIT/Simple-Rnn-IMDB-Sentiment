```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=[
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'i am a good boy',
    'i am a good developer'
    'understand the meaning of the word',
    'your videos are good'
]
sent
```

```python
voc_size=10000
```

```python
one_hot_repr=[one_hot(words,voc_size)for words in sent]
one_hot_repr
```
